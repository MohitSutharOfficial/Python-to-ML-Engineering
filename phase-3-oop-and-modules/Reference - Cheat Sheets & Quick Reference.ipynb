## Frequently Confused Concepts

| Confusion | Clarification |
|-----------|--------------|
| `__str__` vs `__repr__` | `__str__` = pretty for users; `__repr__` = precise for developers. Always define `__repr__`. |
| `classmethod` vs `staticmethod` | `classmethod` gets `cls` (can create instances); `staticmethod` gets neither (pure utility). |
| Inheritance vs Composition | Inheritance: IS-A (`Dog IS-A Animal`). Composition: HAS-A (`Car HAS-A Engine`). Prefer composition for flexibility. |
| `raise` vs `raise e` vs `raise e from orig` | `raise` re-raises current; `raise e` raises with e; `raise new from orig` chains with context. |
| `except Exception` vs `except:` | Always use `except Exception` — bare `except:` also catches `SystemExit` and `KeyboardInterrupt`. |
| `self.x` vs `cls.x` vs `ClassName.x` | `self.x` = instance; `cls.x` inside classmethod = class; `ClassName.x` = class from anywhere. |
| `with open()` vs `open()` | Always use `with open()`. It guarantees file is closed even if an exception occurs. |
