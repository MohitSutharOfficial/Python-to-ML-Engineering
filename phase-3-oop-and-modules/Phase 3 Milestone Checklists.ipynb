## Debugging Mindset for OOP Code

```python
# When AttributeError hits: see all attributes and methods
print(dir(obj))

# When a class behaves unexpectedly: see all instance variables
print(vars(obj))

# When inheritance is confusing: see full method resolution order
print(ClassName.__mro__)

# For exception debugging: get both type AND message
try:
    risky()
except Exception as e:
    print(type(e).__name__, e)

# For file errors: always check path before opening
from pathlib import Path
p = Path('myfile.csv')
print(p.absolute())   # print absolute path for debugging
print(p.exists())     # check before open()
```
