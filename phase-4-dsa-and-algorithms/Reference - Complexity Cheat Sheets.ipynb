## Frequently Confused Concepts

| Confusion | Clarification |
|-----------|--------------|
| BFS vs DFS | BFS uses Queue → level order, shortest path. DFS uses Stack → deep exploration, cycle detection. |
| `list.pop(0)` vs `deque.popleft()` | `list.pop(0)` is O(n). `deque.popleft()` is O(1). Always use `deque` for Queue. |
| BST vs Binary Tree | All BSTs are binary trees. Not all binary trees are BSTs. BST has the ordering property. |
| Memoisation vs Tabulation | Both are DP. Memo=top-down recursive. Tab=bottom-up iterative. Same result, different approach. |
| Stable vs Unstable sort | Stable: equal elements keep their relative order. Merge=stable. Quick=unstable. |
| `in` on list vs set/dict | `list`: O(n). `set`/`dict`: O(1). **Always use set for membership checks in loops.** |

## How Real AI Engineers Use DSA Daily

| Scenario | DSA Applied |
|----------|-------------|
| Choosing between list and set for membership | O(n) vs O(1) — always set for >100 elements |
| Finding top-K predictions from model output | Heap `heapq.nlargest` — O(n log k) vs sort O(n log n) |
| Deduplicating dataset records | Hash map or set — O(n) |
| Loading and splitting dataset | Array slicing O(1) view, shuffle O(n) |
| Building a feature cache for inference | LRU Cache (OrderedDict) — O(1) get/put |
| Scheduling ML pipeline steps | Topological sort on DAG — O(V+E) |
| Nearest neighbour search in embeddings | Binary search or approximate methods |

## Problem-Solving Framework for Any DSA Question

```
1. Understand the problem:  read twice, identify inputs, outputs, constraints
2. Think brute force first: what is the naive O(n²) or O(n³) solution?
3. Identify the bottleneck: which loop can we optimise?
4. Apply pattern recognition: two-pointer? sliding window? BFS? hash map?
5. Implement cleanly:       one thing at a time, test each helper function
6. Verify edge cases:       empty input, single element, duplicates, negatives
7. Analyse complexity:      time AND space, best AND worst
```
