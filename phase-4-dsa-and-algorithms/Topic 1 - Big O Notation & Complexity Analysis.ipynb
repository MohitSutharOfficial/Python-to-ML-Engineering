# Topic 1: Big O Notation & Complexity Analysis

## 1.1 What Is Big O Notation?

Big O notation is a mathematical way to describe how the **runtime or memory usage** of an algorithm grows as the input size grows. It answers: *if I double the data, does my program take twice as long? Ten times? The same time?*

Big O describes the **worst case** and **ignores constants** — it focuses on the growth pattern, not the exact count.

> 📖 **Real-World Analogy — Searching a Phone Book:**
> - **Linear search O(n):** flip pages one by one. 1000 pages = 1000 flips worst case.
> - **Binary search O(log n):** open middle, go left or right. 1000 pages = **10 flips** worst case.
> - **Hash lookup O(1):** you already know the exact page. 0 flips, always instant.
>
> Big O tells you which phone book strategy your algorithm uses.

### 📊 [VISUAL] Big O Growth Chart

```
COMPLEXITY GROWTH (input n = 10, 100, 1000, 1,000,000):

Notation      Name            n=10    n=100    n=1000    n=1M
─────────────────────────────────────────────────────────────────────

FAST ────────────────────────────────────────────────── SLOW
O(1) < O(log n) < O(n) < O(n log n) < O(n²) < O(2^n)
```

## 1.2 Calculating Time Complexity

````python
# ── O(1) — Constant time
# Operations that take the same time regardless of input size
def get_first(lst):
    return lst[0]          # always ONE operation — O(1)

def is_even(n):
    return n % 2 == 0      # always ONE operation — O(1)


# ── O(n) — Linear time: one loop over the input
def find_max(lst):
    max_val = lst[0]
    for x in lst:          # loop runs n times
        if x > max_val:    # 1 comparison each iteration
            max_val = x
    return max_val         # Total: O(n)


# ── O(n²) — Quadratic time: nested loops over the input
def has_duplicates_slow(lst):
    for i in range(len(lst)):             # outer: n iterations
        for j in range(i+1, len(lst)):    # inner: up to n iterations
            if lst[i] == lst[j]:
                return True
    return False   # Total: O(n²) — AVOID for large inputs


# ── O(log n) — Logarithmic: halves the problem each step
def binary_search(lst, target):
    left, right = 0, len(lst) - 1
    while left <= right:              # runs at most log₂(n) times
        mid = (left + right) // 2
        if   lst[mid] == target: return mid
        elif lst[mid] < target:  left  = mid + 1
        else:                    right = mid - 1
    return -1   # Total: O(log n)


# ── O(n log n) — Linearithmic: divide + merge (merge sort)
# Divide input (log n levels) and merge (n work each level)
# Total: O(n log n) — optimal for comparison-based sorting

# ── Quick demo: compare O(n) vs O(n²) on sizes
import time

def time_it(func, *args):
    start = time.perf_counter()
    func(*args)
    return time.perf_counter() - start

data_small  = list(range(1000))
data_medium = list(range(5000))

t1 = time_it(find_max, data_small)
t2 = time_it(find_max, data_medium)
t3 = time_it(has_duplicates_slow, data_small)
t4 = time_it(has_duplicates_slow, data_medium)

print(f"find_max  O(n):  n=1000 -> {t1:.4f}s  |  n=5000 -> {t2:.4f}s  | ratio: {t2/t1:.1f}x")
print(f"dup_check O(n²): n=1000 -> {t3:.4f}s  |  n=5000 -> {t4:.4f}s  | ratio: {t4/t3:.1f}x")
print("O(n) ratio should be ~5x, O(n²) ratio should be ~25x")
````

## 1.3 Space Complexity

| Algorithm | Time | Space | Notes |
|-----------|------|-------|-------|
| Access array element | O(1) | O(1) | Direct index calculation |
| Linear search | O(n) | O(1) | No extra storage needed |
| Binary search (iterative) | O(log n) | O(1) | Just three pointers |
| Binary search (recursive) | O(log n) | O(log n) | Call stack depth = log n |
| Merge sort | O(n log n) | O(n) | Needs temp array for merging |
| Hash map lookup | O(1) avg | O(n) | Stores all n elements |
| BFS / DFS | O(V+E) | O(V) | V=vertices, E=edges |

### 📊 [VISUAL] Complexity Counting Rules

```
COUNTING RULES (simplification):

1. Drop constants:    O(3n)     -> O(n)     O(500)   -> O(1)
2. Drop lower terms:  O(n² + n) -> O(n²)   O(n + log n) -> O(n)
3. Different inputs:  two separate loops -> O(a + b), NOT O(2n)
4. Nested loops (same input) -> multiply: O(n) × O(n) = O(n²)

IDENTIFY PATTERN:
  One loop over n           -> O(n)
  Loop halves each time     -> O(log n)
  Two nested loops over n   -> O(n²)
  Recursive, halves input   -> O(log n) or O(n log n) with merge
  Recursive, all subsets    -> O(2^n)
```

> 🔵 **[AI/ML]** Big O is the #1 reason NumPy exists. A Python loop over 1M elements is O(n) with large constant (~100ns/op). NumPy does the same in C with SIMD vectorisation: same O(n), but **100× smaller constant**. Transformer attention is **O(n²)** in sequence length — that is WHY processing long documents is expensive and WHY researchers invented linear attention O(n).

> 🎯 **[INTERVIEW]**
> - `list.append()` → O(1) amortised
> - `x in my_list` → O(n) (linear scan)
> - `x in my_set` → O(1) average (hash lookup)
> - `sorted()` → O(n log n) — Timsort: merge sort + insertion sort hybrid
