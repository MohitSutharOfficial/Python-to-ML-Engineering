## 7.1 Trees — Concept and Terminology

### 📊 [VISUAL] Binary Search Tree Structure

```
TREE ANATOMY:

                  [10]        <- ROOT (no parent)
                 /    \
              [5]      [15]   <- INTERNAL NODES
             /   \    /   \
           [3]   [7] [12] [20] <- LEAF NODES (no children)

Terminology:
  Root:    topmost node (no parent)                depth = 0
  Leaf:    node with no children
  Height:  longest path from root to any leaf      here = 2
  Depth:   distance from root to a node

BINARY TREE:  each node has at most 2 children (left, right)
BST property: left subtree < node < right subtree at EVERY node
  -> 5's left(3) < 5 < 5's right(7)       TRUE
  -> 10's left(5) < 10 < 10's right(15)   TRUE
```

## 7.2 BST Implementation

````python
class TreeNode:
    def __init__(self, val):
        self.val   = val
        self.left  = None
        self.right = None
    def __repr__(self): return f'TreeNode({self.val})'


class BST:
    """Binary Search Tree with insert, search, and all traversals."""

    def __init__(self):
        self.root = None

    # ── INSERT — O(log n) avg, O(n) worst (skewed tree) ──────────────────────
    def insert(self, val):
        self.root = self._insert(self.root, val)

    def _insert(self, node, val):
        if node is None:     return TreeNode(val)   # base case: place here
        if val < node.val:   node.left  = self._insert(node.left, val)
        elif val > node.val: node.right = self._insert(node.right, val)
        return node

    # ── SEARCH — O(log n) avg ─────────────────────────────────────────────────
    def search(self, val):
        return self._search(self.root, val)

    def _search(self, node, val):
        if node is None:    return False
        if val == node.val: return True
        if val < node.val:  return self._search(node.left, val)
        return              self._search(node.right, val)

    # ── INORDER: Left -> Root -> Right — produces SORTED output for BST ───────
    def inorder(self):
        result = []
        def _in(node):
            if node:
                _in(node.left)
                result.append(node.val)
                _in(node.right)
        _in(self.root)
        return result

    # ── PREORDER: Root -> Left -> Right ──────────────────────────────────────
    def preorder(self):
        result = []
        def _pre(node):
            if node:
                result.append(node.val)
                _pre(node.left)
                _pre(node.right)
        _pre(self.root)
        return result

    # ── POSTORDER: Left -> Right -> Root ─────────────────────────────────────
    def postorder(self):
        result = []
        def _post(node):
            if node:
                _post(node.left)
                _post(node.right)
                result.append(node.val)
        _post(self.root)
        return result

    # ── LEVEL ORDER: BFS — level by level ────────────────────────────────────
    def level_order(self):
        from collections import deque
        if not self.root: return []
        result, queue = [], deque([self.root])
        while queue:
            node = queue.popleft()
            result.append(node.val)
            if node.left:  queue.append(node.left)
            if node.right: queue.append(node.right)
        return result

    def height(self):
        def _h(node):
            if not node: return 0
            return 1 + max(_h(node.left), _h(node.right))
        return _h(self.root)


bst = BST()
for v in [10, 5, 15, 3, 7, 12, 20]:
    bst.insert(v)

print("Inorder:    ", bst.inorder())      # [3,5,7,10,12,15,20] — sorted!
print("Preorder:   ", bst.preorder())     # [10,5,3,7,15,12,20]
print("Postorder:  ", bst.postorder())    # [3,7,5,12,20,15,10]
print("Level order:", bst.level_order())  # [10,5,15,3,7,12,20]
print("Height:     ", bst.height())       # 3
print("Search 7:   ", bst.search(7))      # True
print("Search 99:  ", bst.search(99))     # False
````

## 7.3 Tree Traversal Patterns

### 📊 [VISUAL] Traversal Order Summary

```
         [10]
        /    \
      [5]    [15]
      / \
    [3] [7]

Inorder   (L-Root-R): 3, 5, 7, 10, 15   <- SORTED for BST
Preorder  (Root-L-R): 10, 5, 3, 7, 15   <- root first (tree copy)
Postorder (L-R-Root): 3, 7, 5, 15, 10   <- leaves first (delete safely)
Level Order (BFS):    10, 5, 15, 3, 7   <- layer by layer

MEMORY TRICK:
  Pre  = ROOT before children  (Pre  = before)
  In   = ROOT between children (In   = between)
  Post = ROOT after children   (Post = after)
```

> 🔵 **[AI/ML]** Decision Trees in sklearn: internal node = feature split, leaf = predicted class. The tree structure IS a BST variant. PyTorch computation graph is a DAG. **Backpropagation traverses this graph in postorder** — leaves (inputs) first, root (loss) last. Random Forests = ensemble of decision trees. Gradient Boosting Trees (XGBoost, LightGBM) = highest-performing ML algorithm for tabular data.
