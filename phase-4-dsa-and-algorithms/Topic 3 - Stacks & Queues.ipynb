## 3.1 Stack — Last In First Out (LIFO)

A stack is a collection where you can **only add and remove from one end** (the top). The last item added is always the first item removed — like a stack of plates.

### 📊 [VISUAL] Stack Memory Model

```
STACK — LIFO (Last In First Out)

      PUSH(10)  PUSH(20)  PUSH(30)     POP()     POP()
                                       returns   returns
 __    __         __        [30] <-TOP   30        20
 ||    |10|       |10|      [20]        __         __
 ||    |__|       |20|      [10]       |10|       |10|
empty  1 item    2 items   3 items    1 item    1 item

Operations:   push(x) O(1)   pop() O(1)   peek() O(1)
              is_empty() O(1)   len() O(1)

REAL USES:
  - Browser back button (undo history)
  - Function call stack (Python's own call stack IS a stack)
  - Balanced parentheses validation
  - DFS traversal (iterative)
  - Expression evaluation
```

````python
class Stack:
    """Stack (LIFO) — implemented with Python list."""

    def __init__(self, max_size=None):
        self._data     = []
        self._max_size = max_size

    def push(self, val):            # O(1)
        if self._max_size and len(self._data) >= self._max_size:
            raise OverflowError(f'Stack is full (max={self._max_size})')
        self._data.append(val)

    def pop(self):                  # O(1)
        if self.is_empty():
            raise IndexError('Pop from empty stack')
        return self._data.pop()

    def peek(self):                 # O(1)
        if self.is_empty():
            raise IndexError('Peek at empty stack')
        return self._data[-1]

    def is_empty(self): return len(self._data) == 0
    def __len__(self):  return len(self._data)
    def __repr__(self): return f'Stack(top->{self._data[::-1]})'


# ── Classic stack problem: balanced parentheses ───────────────────────────────
def is_balanced(s):
    stack  = Stack()
    pairs  = {')':'(', ']':'[', '}':'{'}
    opener = set(pairs.values())
    for ch in s:
        if ch in opener:
            stack.push(ch)
        elif ch in pairs:
            if stack.is_empty() or stack.pop() != pairs[ch]:
                return False
    return stack.is_empty()


print(is_balanced('({[]})'))    # True
print(is_balanced('({[})'))     # False
print(is_balanced('((()))'))    # True
print(is_balanced('([)]'))      # False

# ── Stack demo ────────────────────────────────────────────────────────────────
s = Stack(max_size=5)
for v in [1, 2, 3]: s.push(v)
print(s)
print("peek:", s.peek())   # 3
print("pop:", s.pop())     # 3
print(s)
````

## 3.2 Queue — First In First Out (FIFO)

### 📊 [VISUAL] Queue Memory Model

```
QUEUE — FIFO (First In First Out)

ENQUEUE: add to REAR          DEQUEUE: remove from FRONT

FRONT                                   REAR
 [10] <- dequeue removes here  enqueue adds here -> [30]
 [10] -> [20] -> [30]

Like a checkout line: person who joins first gets served first.

REAL USES:
  - BFS (Breadth-First Search) traversal
  - Task scheduling (OS process queue)
  - ML: Experience Replay Buffer in Reinforcement Learning (fixed-size queue)
  - Data stream buffering
```

> 🔵 **[AI/ML]** Experience Replay Buffer in RL IS a fixed-size queue:
> `replay_buffer = deque(maxlen=100_000)` — stores `(state, action, reward, next_state)` tuples.
> Priority Queue is used in Dijkstra's (knowledge graphs) and Beam Search in NLP.

````python
from collections import deque
import heapq


class Queue:
    """Queue (FIFO) — uses deque for O(1) front removal."""

    def __init__(self, max_size=None):
        self._data     = deque()
        self._max_size = max_size

    def enqueue(self, val):         # O(1)
        if self._max_size and len(self._data) >= self._max_size:
            raise OverflowError('Queue is full')
        self._data.append(val)

    def dequeue(self):              # O(1) — the key advantage over list!
        if self.is_empty():
            raise IndexError('Dequeue from empty queue')
        return self._data.popleft()

    def front(self):    return self._data[0]
    def is_empty(self): return len(self._data) == 0
    def __len__(self):  return len(self._data)
    def __repr__(self): return f'Queue({list(self._data)})'


# WHY deque over list:
# list.pop(0)      -> O(n) shifts all elements left
# deque.popleft()  -> O(1) doubly-linked list under the hood

q = Queue()
for v in [10, 20, 30]: q.enqueue(v)
print(q)
print("dequeue:", q.dequeue())   # 10 — FIFO
print(q)


# ── Priority Queue (min-heap) ─────────────────────────────────────────────────
class PriorityQueue:
    """Min-heap based priority queue."""
    def __init__(self):
        self._heap = []

    def push(self, priority, item):   # O(log n)
        heapq.heappush(self._heap, (priority, item))

    def pop(self):                    # O(log n)
        return heapq.heappop(self._heap)

    def __len__(self):  return len(self._heap)
    def __repr__(self): return f'PQ({self._heap})'


pq = PriorityQueue()
pq.push(3, 'low priority task')
pq.push(1, 'urgent task')
pq.push(2, 'medium task')
print("\nPriority Queue pops:")
while pq:
    print(" ", pq.pop())   # (1, urgent), (2, medium), (3, low)
````
