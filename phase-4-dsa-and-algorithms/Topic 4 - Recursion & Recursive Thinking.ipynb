## 4.1 What Is Recursion?

Recursion is when a **function calls itself** to solve a smaller version of the same problem. Every recursive solution has two parts:
1. **Base case** — when to stop (the answer we know directly)
2. **Recursive case** — how to break the problem smaller

> 🪆 **Real-World Analogy — Russian Nesting Dolls (Matryoshka):**
> To find the smallest doll: *Open the doll. Is there a smaller one inside? (recursive) Is there no smaller one? (base case — stop, return this doll).* Each step does the same action on a smaller version.

### 📊 [VISUAL] Recursion Call Stack

```
RECURSION ANATOMY:

def factorial(n):
    if n == 0:                   <- BASE CASE: stop condition
        return 1                    answer we know without recursion
    return n * factorial(n-1)    <- RECURSIVE CASE: smaller subproblem

CALL STACK for factorial(4):
  factorial(4) -> 4 * factorial(3)
                      3 * factorial(2)
                          2 * factorial(1)
                              1 * factorial(0)
                                  returns 1     <- BASE CASE hit
                              returns 1*1 = 1
                          returns 2*1 = 2
                      returns 3*2 = 6
  returns 4*6 = 24

Stack GROWS with each call. Stack SHRINKS as calls return.
Max depth = n. Space: O(n) for call stack.
```

## 4.2 Recursion Examples — Beginner to Advanced

````python
# ── BEGINNER: Factorial — O(n) ──────────────────────────────────────────────
def factorial(n):
    if n == 0: return 1          # base case
    return n * factorial(n - 1)  # recursive case

print("5! =", factorial(5))   # 120


# ── FIBONACCI: naive O(2^n) vs memoised O(n) ─────────────────────────────────
def fib_naive(n):
    if n <= 1: return n
    return fib_naive(n-1) + fib_naive(n-2)   # two recursive calls
# DANGEROUS: fib_naive(40) takes ~30 seconds!

# Memoised — top-down dynamic programming
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n <= 1: return n
    return fib(n-1) + fib(n-2)
# fib(100) is instant — each subproblem computed only ONCE

print("fib(10) =", fib(10))   # 55
print("fib(50) =", fib(50))   # 12586269025 — instant with cache


# ── POWER: O(log n) via divide & conquer ─────────────────────────────────────
def power(base, exp):
    if exp == 0:         return 1
    if exp % 2 == 0:
        half = power(base, exp // 2)
        return half * half           # square the half result — halves exp!
    return base * power(base, exp - 1)

print("2^10 =", power(2, 10))   # 1024


# ── MERGE SORT: O(n log n) — classic divide & conquer ────────────────────────
def merge_sort(arr):
    if len(arr) <= 1: return arr       # base case
    mid   = len(arr) // 2
    left  = merge_sort(arr[:mid])      # sort left half
    right = merge_sort(arr[mid:])      # sort right half
    return _merge(left, right)          # merge sorted halves

def _merge(left, right):               # O(n) merge step
    result = []
    i = j  = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]: result.append(left[i]);  i += 1
        else:                   result.append(right[j]); j += 1
    return result + left[i:] + right[j:]

print("merge sort:", merge_sort([38, 27, 43, 3, 9, 82, 10]))


# ── BACKTRACKING: generate all subsets — O(2^n) ──────────────────────────────
def subsets(nums):
    result = []
    def backtrack(start, current):
        result.append(list(current))    # add current subset
        for i in range(start, len(nums)):
            current.append(nums[i])
            backtrack(i + 1, current)   # recurse with next element
            current.pop()               # undo choice (backtrack)
    backtrack(0, [])
    return result

print("subsets([1,2,3]):", subsets([1, 2, 3]))
````

## 4.3 Recursion vs Iteration — When to Use Which

| Situation | Use Recursion | Use Iteration |
|-----------|--------------|---------------|
| Problem is naturally recursive | Trees, graphs, divide & conquer | Simple loops, counters |
| Readability matters | Tree traversal is cleaner recursive | Summation is cleaner iterative |
| Stack depth concern | Avoid: Python default limit ~1000 | Safe for any depth |
| Performance critical | Adds function call overhead | Slightly faster (no call overhead) |
| Memory concern | O(depth) call stack | O(1) extra if in-place |

> 💡 **[PRO TIP]** Python's default recursion limit is 1000. For deep recursion: `import sys; sys.setrecursionlimit(10000)`. Use `@lru_cache(maxsize=None)` to memoize recursive functions. **Always identify your base case FIRST.**

> 🎯 **[INTERVIEW]**
> - `RecursionError` cause → missing or unreachable base case
> - Converting recursion → iteration → use an explicit stack data structure
> - `fib_naive(n)` complexity → O(2^n)
> - Memoisation vs Tabulation → memo = top-down recursive; tab = bottom-up iterative
