## Data Structures at a Glance

| Structure | Syntax | Mutable | Key Feature |
|-----------|--------|---------|-------------|
| List | `[1, 2, 3]` | ✅ Yes | Ordered, index access, append/remove |
| Tuple | `(1, 2, 3)` | ❌ No | Immutable, unpack, dict-key safe |
| Dictionary | `{k: v}` | ✅ Yes | Key lookup, named fields, nesting |
| Set | `{1, 2, 3}` | ✅ Yes | Unique, fast lookup, set math |
